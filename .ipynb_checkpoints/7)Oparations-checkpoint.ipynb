***DataFrames Applying Functions***

```python
df1
```

```python
def square(x):
    return x**2 # Column element squire like 10^10, 20^20, 30^30 etc
```

```python
df1["D"] = df1["B"].apply(square) # Apply on Column B and add NEW col it D
```

```python
df1
```

```python
df1['D'] = df1["B"].apply(lambda x : x**2)
df1
```
