# Pandas Series

A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

### Creating Series
There are multiple ways to create a pandas Series:

```python
import numpy as np
import pandas as pd
```

```python
labels = ["a", "b", "c"]
my_list = [10, 20, 30]
arr = np.array([10,20,30])
dic = {1:10, 2:20, 3:30}
```

```python
# now create a series
pd.Series(my_list)
```

```python
# set custom index/labels 
pd.Series(my_list, index=labels)
```

```python
pd.Series(dic)
```

```python
pd.Series(arr)
```

```python
print("series done")
```
