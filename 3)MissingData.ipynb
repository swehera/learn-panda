```python
import numpy as np 
import pandas as pd 
```

#### Finding Missing Data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

#### Removing Missing Data

```python
df
```

```python
df.dropna() # only given the row where no nan value exits
```

```python
df
```

```python
df.dropna(thresh=3) # thresh=3 mean return remove rows where not atleat 3 elements not exists
```

```python
df.dropna(thresh=4) # means retun row where 4 elements are exits and all rows are removed
```

#### Filling The Missing Data

```python
df
```

```python
df.fillna(0) # means fill all the NaN elements with 0 value
```

```python
# now we can also fill the NaN value with our custom Value
new_value = {"A": 0, "B":100, "C":300, "D": 500}
df.fillna(value=new_value)
```

```python
df
```

```python
df.fillna(df.mean())
```
